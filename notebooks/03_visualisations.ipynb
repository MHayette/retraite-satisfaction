# 📈 Notebook 03 — Visualisations
**Objectif** : Histogrammes, boxplots, heatmap, courbes d'évolution
Tous les graphiques sont sauvegardés dans `reports/figures/`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
df = pd.read_csv('../data/processed/retraite_clean.csv')

COLS_SERVICES = ['Accueil', 'Soins_Qualite', 'Competence_Personnel',
    'Disponibilite_Personnel', 'Reconfort', 'Qualite_Restauration',
    'Hygiene', 'Confort_Chambre', 'Services_Annexes']

print('✅ Données chargées')
```

## 1. Histogramme — Distribution de la Satisfaction

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(df['Satisfaction'], bins=10, color='steelblue', edgecolor='white', density=False)
ax.axvline(df['Satisfaction'].mean(), color='red', linestyle='--', linewidth=2,
           label=f'Moyenne : {df["Satisfaction"].mean():.1f}')
ax.axvline(df['Satisfaction'].median(), color='orange', linestyle='--', linewidth=2,
           label=f'Médiane : {df["Satisfaction"].median():.1f}')
ax.set_title('Distribution de la Satisfaction globale (sur 10)', fontsize=13, fontweight='bold')
ax.set_xlabel('Note de satisfaction')
ax.set_ylabel('Nombre de résidents')
ax.legend()
plt.tight_layout()
plt.savefig('../reports/figures/satisfaction_globale.png', dpi=150)
plt.show()
```

## 2. Boxplots — Satisfaction par niveau d'autonomie (Public)

```python
fig, ax = plt.subplots(figsize=(8, 5))
df.boxplot(column='Satisfaction', by='Public', ax=ax,
           patch_artist=True,
           boxprops=dict(facecolor='lightblue', color='steelblue'),
           medianprops=dict(color='red', linewidth=2))
ax.set_title('Satisfaction selon le type de Public', fontsize=13, fontweight='bold')
ax.set_xlabel('Type de Public')
ax.set_ylabel('Satisfaction (sur 10)')
plt.suptitle('')
plt.tight_layout()
plt.savefig('../reports/figures/boxplot_public.png', dpi=150)
plt.show()
print('Interprétation : Les résidents Valides affichent une satisfaction inférieure aux autres groupes')
```

## 3. Boxplot — Prix de la pension par Formule

```python
fig, ax = plt.subplots(figsize=(9, 5))
df.boxplot(column='PRIX_PENSION', by='Formule', ax=ax,
           patch_artist=True,
           boxprops=dict(facecolor='lightyellow', color='goldenrod'),
           medianprops=dict(color='red', linewidth=2))
ax.set_title('Prix de la pension par Formule', fontsize=13, fontweight='bold')
ax.set_xlabel('Formule')
ax.set_ylabel('Prix de la pension')
plt.suptitle('')
plt.tight_layout()
plt.savefig('../reports/figures/boxplot_prix_formule.png', dpi=150)
plt.show()
```

## 4. Scatter plot — Prix vs Satisfaction

```python
from numpy.polynomial.polynomial import polyfit

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df['PRIX_PENSION'], df['Satisfaction'], alpha=0.4, color='darkgreen', s=30)

# Droite de régression
m, b = np.polyfit(df['PRIX_PENSION'], df['Satisfaction'], 1)
x_line = np.linspace(df['PRIX_PENSION'].min(), df['PRIX_PENSION'].max(), 100)
ax.plot(x_line, m * x_line + b, color='red', linewidth=2, label=f'Régression (r≈0.095)')

ax.set_title('Relation entre Prix et Satisfaction', fontsize=13, fontweight='bold')
ax.set_xlabel('Prix de la pension')
ax.set_ylabel('Satisfaction (sur 10)')
ax.legend()
plt.tight_layout()
plt.savefig('../reports/figures/prix_vs_satisfaction.png', dpi=150)
plt.show()
print('Conclusion : Droite quasi horizontale → le prix n\'influence pas la satisfaction (r=0.095)')
```

## 5. Heatmap — Matrice de corrélations

```python
cols_corr = COLS_SERVICES + ['PRIX_PENSION', 'Satisfaction']
corr_matrix = df[cols_corr].corr()

fig, ax = plt.subplots(figsize=(12, 9))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, ax=ax, square=True, linewidths=0.5)
ax.set_title('Matrice de corrélations', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('../reports/figures/heatmap_correlations.png', dpi=150)
plt.show()
```

## 6. Bar chart — Satisfaction moyenne par service (classement)

```python
moyennes = df[COLS_SERVICES].mean().sort_values()
moyenne_globale = moyennes.mean()

fig, ax = plt.subplots(figsize=(10, 6))
colors = ['#e74c3c' if v < moyenne_globale else '#2980b9' for v in moyennes]
bars = ax.barh(moyennes.index, moyennes.values, color=colors, edgecolor='white', height=0.6)
ax.axvline(moyenne_globale, color='black', linestyle='--', linewidth=1.5,
           label=f'Moyenne : {moyenne_globale:.2f}/5')

for bar, val in zip(bars, moyennes.values):
    ax.text(val + 0.02, bar.get_y() + bar.get_height()/2,
            f'{val:.2f}', va='center', fontsize=10)

ax.set_title('Classement des services par satisfaction moyenne (sur 5)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Note moyenne')
ax.set_xlim(0, 5.5)
ax.legend()
plt.tight_layout()
plt.savefig('../reports/figures/classement_services.png', dpi=150)
plt.show()
```
